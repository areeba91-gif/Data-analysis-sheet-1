```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

```

```python
ecom = pd.read_csv("pakistani_ecommerce_dataset.csv")
print(ecom.shape)
print(ecom.isna().sum())
```

```python
print("Duplicate rows:", ecom.duplicated().sum())
ecom.head()
```

```python
print(ecom["product_category"].value_counts())
print("Most popular:", ecom["product_category"].value_counts().idxmax())
```

```python
print("Average order value:", round(ecom["order_value"].mean(), 2))
print("Median order value:", ecom["order_value"].median())
```

```python
print(ecom.groupby("order_status")["order_value"].agg(["count", "mean"]).round(2))
```

```python
print(ecom["city"].value_counts())
print("Most orders:", ecom["city"].value_counts().idxmax())
```

```python
print(ecom["payment_method"].value_counts())
print("Most used:", ecom["payment_method"].value_counts().idxmax())
```

```python
avg_city = ecom.groupby("city")["order_value"].mean().sort_values(ascending=False)
print(avg_city.round(2))

plt.figure(figsize=(8, 5))
plt.bar(avg_city.index, avg_city.values)
plt.title("Average Order Value by City")
plt.xlabel("City")
plt.ylabel("Average order value (PKR)")
plt.show()
```

```python
student = pd.read_csv("student_lifestyle_dataset.csv")
print(student.shape)
print(student.isna().sum())
```

```python
print("Duplicate rows:", student.duplicated().sum())
student.head()
```

```python
print("Average study hours:", round(student["study_hours"].mean(), 2))
print("Median study hours:", student["study_hours"].median())
```

```python
print(student.groupby("gender")["sleep_hours"].agg(["count", "mean", "median"]).round(2))
```

```python
r = student["screen_time"].corr(student["sleep_hours"])
print("Correlation:", round(r, 3))

plt.figure(figsize=(7, 5))
plt.scatter(student["screen_time"], student["sleep_hours"])
plt.title("Screen Time vs Sleep Hours")
plt.xlabel("Screen time")
plt.ylabel("Sleep hours")
plt.show()
```

```python
print(student["daily_transport_cost"].describe().round(2))

plt.figure(figsize=(8, 5))
plt.hist(student["daily_transport_cost"], bins=10, edgecolor="black")
plt.title("Daily Transport Cost of Students")
plt.xlabel("Daily transport cost")
plt.ylabel("Number of students")
plt.show()
```

```python
internet = student.groupby("gender")["internet_usage"].mean()
print(internet.round(2))

plt.figure(figsize=(6, 5))
plt.bar(internet.index, internet.values)
plt.title("Average Internet Usage by Gender")
plt.xlabel("Gender")
plt.ylabel("Average internet usage")
plt.show()
```
